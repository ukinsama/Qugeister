```python
from graphviz import Digraph

def visualize_cqcnn_structure_with_clusters(cnn_layers=None, qnn_layers=None, title="CQ-CNN Hybrid Model"):
    """
    CNNを上段、QNNを下段に分けた2段構成のクラスタ図。
    """
    dot = Digraph(comment=title)
    dot.attr(rankdir='TB')  # 上から下方向（縦配置）

    # --- CNN クラスタ ---
    with dot.subgraph(name='cluster_cnn') as c:
        c.attr(style='filled', color='lightblue', label='CNN')
        c.node("Input", "Input", shape="oval", style="filled", color="lightgray")
        prev = "Input"

        if cnn_layers is None:
            cnn_layers = ["CNN Feature Extractor"]
        for idx, layer in enumerate(cnn_layers):
            node_id = f"CNN{idx}"
            c.node(node_id, layer)
            c.edge(prev, node_id)
            prev = node_id

        c.node("CNN_out", "CNN Output")
        c.edge(prev, "CNN_out")
        prev = "CNN_out"

    # --- QNN クラスタ ---
    with dot.subgraph(name='cluster_qnn') as q:
        q.attr(style='filled', color='lightpink', label='QNN')
        q.node("QNN_input", "QNN Input")
        q.edge("CNN_out", "QNN_input", constraint='false')  # 横方向の接続（位置を崩さない）
        prev_q = "QNN_input"

        if qnn_layers is None:
            qnn_layers = ["AngleEmbedding", "RealAmplitudes", "Expval"]
        for idx, layer in enumerate(qnn_layers):
            node_id = f"QNN{idx}"
            q.node(node_id, layer)
            q.edge(prev_q, node_id)
            prev_q = node_id

    # 出力
    dot.node("Output", "Output", shape="oval", style="filled", color="lightgray")
    dot.edge(prev_q, "Output")

    # rank 制御: CNNを上段、QNNを下段
    with dot.subgraph() as s:
        s.attr(rank='same')
        s.node("Input")
        s.node("CNN0")
        s.node("CNN_out")
    # QNN系は自動的にその下に配置される

    # グラフ属性
    dot.attr(label=title, fontsize='20', fontcolor='black')
    dot.attr(labeljust='c', labelloc='t')
    dot.attr('node', shape='box', style='filled', color='lightyellow')
    dot.attr('edge', color='black')

    return dot

# 使用例
dot = visualize_cqcnn_structure_with_clusters(
    cnn_layers=["Conv2d"],
    qnn_layers=["AngleEmbedding", "RealAmplitudes"]
)
dot.render("cqcnn_model_clustered_graph", format="png", view=True)
```
