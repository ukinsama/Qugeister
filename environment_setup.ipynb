## PennyLaneデバイス動作確認

```python

import pennylane as qml

dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def circuit():
    qml.Hadamard(0)
    qml.CNOT([0, 1])
    return qml.state()

print("Sample QNN output:", circuit())
```
