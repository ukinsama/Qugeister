
# 対戦・棋譜再生チュートリアル

このノートブックでは、**学習済みモデルの対戦**と**棋譜（対戦ログ）の再生**を行います。
- ランダムポリシーとの対戦例
- 対戦ログの生成とJSON保存
- 棋譜を1手ずつ再生表示


## 必要ライブラリのインポート

```python
import os
import json
```

```python
from geister_game import GeisterGame
from train import Env_Geister, RandomPolicy
```

## エージェントの準備（例：ランダム vs ランダム）

```python

game = GeisterGame()
agentA = RandomPolicy("A", game)
agentB = RandomPolicy("B", game)
env = Env_Geister(agentA, agentB, game)
print("✅ エージェントと環境を準備しました。")
```

## 1試合の対戦とログ保存

```python

winner, moves_log, final_board = env.play_one_game_with_log()

os.makedirs("logs/tutorial", exist_ok=True)
log_path = "logs/tutorial/sample_game.json"
with open(log_path, "w", encoding="utf-8") as f:
    json.dump({
        "winner": winner,
        "moves": moves_log,
        "final_board": final_board
    }, f, indent=2)

print(f"対戦結果: Winner = {winner}")
print(f"棋譜ログを {log_path} に保存しました。")
```

## 棋譜再生（1手ずつ盤面表示）

```python

import time

print("=== 棋譜再生開始 ===")
game = GeisterGame()
game.display_board(reveal_opponent_pieces=True)

for move in moves_log:
    time.sleep(0.5)
    print(f"Turn {move['turn']} - Player {move['player']} moves {move['action']}")
    game.make_move(move['action'][0], move['action'][1])
    game.display_board(reveal_opponent_pieces=True)

print("=== 棋譜再生終了 ===")
```
